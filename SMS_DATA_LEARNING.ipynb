```python
spam_msgs = df[df['v1'] == 'spam']

for i, msg in enumerate(spam_msgs['v2'].head(20), 1):
    print(f"\nSpam {i}:")
    print(msg)
```

```python
ham_msgs = df[df['v1'] == 'ham']

for i, msg in enumerate(ham_msgs['v2'].head(20), 1):
    print(f"\nHam {i}:")
    print(msg)
```

```python
#COMMON PATTERNS IN SPAM MESSAGES
1. FREE REWARDS: Many spam messages offer free rewards, prizes, or gifts to entice recipients to click on links or provide personal information.
2. URGENCY AND THREATS: Spam messages often create a sense of urgency or use threats to pressure recipients into taking immediate action, such as claiming a prize or avoiding account suspension.
3. SUSPICIOUS LINKS: Spam messages frequently contain links to suspicious websites or ask recipients to click on links that may lead to phishing sites or malware downloads.
4. REQUEST FOR PERSONAL INFORMATION: Spam messages may ask for sensitive personal information, such as passwords, credit card details, or social security numbers, which legitimate organizations would not request via SMS.
5. POOR GRAMMAR AND SPELLING: Many spam messages contain noticeable grammar and spelling errors, which can be a red flag for recipients.
6. UNFAMILIAR SENDER: Spam messages often come from unknown or suspicious phone numbers, email addresses, or shortcodes that recipients do not recognize.
```

```python
#COMMON PATTERNS IN HAM MESSAGES
1. PERSONALIZED GREETINGS: Ham messages often include personalized greetings, such as the recipient 's name or a friendly salutation, which indicates a genuine sender.
2. RELEVANT CONTENT: Ham messages typically contain content that is relevant to the recipient, such as updates from friends, family, or work-related information, rather than generic or unsolicited offers.
3. CLEAR INTENT: Ham messages usually have a clear purpose or intent, such as providing information, asking a question, or making a request, without using manipulative tactics or pressure.    
4. APPROPRIATE LANGUAGE: Ham messages generally use appropriate language and tone, avoiding excessive capitalization, exclamation marks, or aggressive language that is often found in spam messages.
5. TRUSTED SENDER: Ham messages are often sent from known contacts, such as friends, family members, or reputable organizations, which helps establish trust and credibility.
```
